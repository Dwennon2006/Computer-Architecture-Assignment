```python
def to_binary(n):
    return format(n & 0xFF, '08b')


def signed_binary_addition(a, b):
    result = a + b

    print("\nExpression:", a, "+", b)
    print("First operand:", to_binary(a))
    print("Second operand:", to_binary(b))

    print("\nBinary Addition:")
    print(" ", to_binary(a))
    print("+", to_binary(b))
    print("----------")
    print(" ", to_binary(result))

    print("\nBinary Result:", to_binary(result))
    print("Decimal Result:", result)

    if result < -128 or result > 127:
        print("Arithmetic Overflow: Yes")
    else:
        print("Arithmetic Overflow: No")

    print("-" * 35)


# a) 25 + (+18)
signed_binary_addition(25, 18)

# b) (-25) + (-18)
signed_binary_addition(-25, -18)
```
